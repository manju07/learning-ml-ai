```python

# Activity 1: Create a DataFrame from a dictionary and display it
activity1_data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, 92, 78]
}
df1 = pandas.DataFrame(activity1_data)
print("Activity 1 - DataFrame from dict:")
print(df1, end="\n\n")

# Activity 2: Read a column as a list
names_list = df1['Name'].tolist()
print("Activity 2 - Names column as list:", names_list, end="\n\n")

# Activity 3: Add a new column based on existing data
df1['Passed'] = df1['Score'] >= 80
print("Activity 3 - DataFrame with new column 'Passed':")
print(df1, end="\n\n")

# Activity 4: Filter rows where Score >= 90
high_scorers = df1[df1['Score'] >= 90]
print("Activity 4 - Students with Score >= 90:")
print(high_scorers, end="\n\n")

# Activity 5: Describe summary statistics
print("Activity 5 - Summary statistics:")
print(df1.describe(), end="\n\n")


```
